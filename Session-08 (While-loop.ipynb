## ***While Loop in python*** 

Keep looping until certain condition remains true

```python
#             For Loop                                        	While Loop
# Sequence ya items ke hisaab se chalta hai.	           Condition ke hisaab se chalta hai.
# Items complete hone par stop hota hai.                 Condition False hone par stop hota hai.
# Jab iterations ka pata ho, use karte hain.             Jab condition ke basis par loop chalana ho, use karte hain.


# for loop + range → Python khud sequence generate karta hai aur values ek-ek karke deta hai. sequence sy bana kar deta ha 
# while loop → Aapko khud sequence banana padta hai (start, condition, update).
```

```python
x = 0
x <= 5 ## condi true while true condit (condi false infinite loop)
```

```python
x = 0

while x <= 5:  
    print(x)  
    x = x + 1

# while → keyword, used to start a while loop.

# x <= 5 → condition. The loop will keep running as long as this condition is True.(conditioon false ho jye loop break ho jye)

# x → loop variable, it starts at 1 and changes every time.

# x = x + 1 → update step, increases the value of x so that the loop can eventually stop.

# First check: x = 1 → condition 1 <= 3 is True → print 1.
# Then x = 2 → condition 2 <= 3 is True → print 2.
# Then x = 3 → condition 3 <= 3 is True → print 3.
# Then x = 6 → condition 6 <= 5 is False → loop stops.
```

                                
## ***NESTED WHILE LOOP*** 

```python
week_days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
time_slots = ['9AM-12PM', '12PM-03PM', '03PM-06PM',]

i = 0

while i < len(week_days):
    print(week_days[i])
    i += 1            # 1st loop end hoa

    j = 0             # 2nd True cond mli wo run ki
    while j < len(time_slots):
        print(time_slots[j])
        j += 1          # j end then again move to i=0
    print('')
```

Question 5:

Print total amount spent by each customer (Using nested while loop)

orders = [ {"order_id": 1, "customer": "Ali", "items": [120, 80, 50]}, {"order_id": 2, "customer": "Sara", "items": [200, 150]}, {"order_id": 3, "customer": "Raheel", "items": [90, 60, 40, 30]}, {"order_id": 4, "customer": "Naveed", "items": [300]}, ]

Expected Output:

Ali spent total: 250

Sara spent total: 350

Raheel spent total: 220

Naveed spent total: 300

```python
orders = [ {"order_id": 1, "customer": "Ali", "items": [120, 80, 50]},
          {"order_id": 2, "customer": "Sara", "items": [200, 150]}, 
          {"order_id": 3, "customer": "Raheel", "items": [90, 60, 40, 30]},
          {"order_id": 4, "customer": "Naveed", "items": [300]}, ]
orders  # list in dict
```

```python
x = 0
while x < len(orders):
    items = orders[x]["items"]
    y = 0
    total = 0

    while y < len(items):
        total += items[y]
        y += 1

    print(f"{orders[x]['customer']} spent total: {total}")
    x += 1
```

```python
# write a program to print countdown reverse order if user input random number

# Example
# Enter number: 5

# output:
# 5
# 4
# 3
# 2
# 1
# Time up

reverse_Num = int(input('Enter a Reverse Number to Start Countdown'))

while reverse_Num > -0:
    print(reverse_Num)
    reverse_Num -= 1

print('Time Up!')
```

### ***While Loop (Continue)*** 

```python
i = 0
while i < 5:
    i += 1
    if i == 3:
        continue  # 3 ko skip kar dega, neeche wala print nahi chalega
    print(i)
```

```python
num = 0
while num < 10:
    num += 1
    if num % 2 == 0:  # Agar number 2 se divide ho (Even ho)
        continue      # Toh use skip kar do
    print(num)
```

### ***While Loop (Break)*** 

```python
i = 1
while i <= 10:
    if i == 5:
        print("5 mil gaya, loop band!")
        break  # Loop yahin khatam ho jayega, 6 se 10 print nahi honge
    print(i)
    i += 1
```
