# ***NESTED for LOOPS*** 

```python

week_days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday']
time_slots = ['9AM-10AM', '10AM-11AM', '11AM-12PM']
```

```python
for weekDay in week_days:
    print('-- ',weekDay) # only print week days
```

```python
for timeSlots in time_slots:
    print(f'- {timeSlots}') 
```

```python
for weekDay in week_days:
    for timeSlots in time_slots:
        print(weekDay, timeSlots)
```

```python
## NESTED FOR LOOP

week_days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday']
time_slots = ['9AM-10AM', '10AM-11AM', '11AM-12PM']
```

```python
for weekDay in week_days:  # Outer loop
    print(weekDay)
    for timeSlots in time_slots:  # inner loop
        print(timeSlots)
    print('')             # # Outer loop
```

```python
for weekDay in week_days:  # Outer loop
    print(weekDay)
    for timeSlots in time_slots:  # inner loop
        print(timeSlots)

# phly outer run hoga then inner 
# inner run all hokar then again outer pr move krega 
```

#### ***reverse counting for loop- range***

```python
for reverse_count in range(8, 0, -1):

    print(f'{reverse_count}', ' <<=  <= reverse')
    
# start 8, end 0 exclude 1 tk, -1 reverse
```

# ***Continue*** 

```python
for a in range(1, 6):
    if a == 3:
        continue #skip
```

```python
for j in range(4, 11):

    if j == 8:
        continue  # skip 8 no

    else:
        print(j)
```

# ***Break*** 

```python
for u in range(-5, 5):
    # print(u)
    if u == 3:
        break
    else:
        print(u)
```
